## 2023weather.ipynb
2023 날씨데이터 머신러닝

```python
import pandas as pd
weather = pd.read_excel('weather2023.xlsx', index_col='date')
weather.head() 
```

```python
print(weather.isnull().sum())
```

```python
weather.drop(['rain', 'location', 'loc_name'], axis=1, inplace=True)
weather.head()
```

```python
weather.info()
```

```python
weather.describe()
```

describe() : 데이터 프레임이나 특정열의 기초
통계갓은 한번에 확인하는 매서드, 각열의
개수, 평균, 표준편차, 최솟값, 사분위소,
최대값을 보여줌.

```python
weather.corr()
```

corr() : 열과 열 사이의 상관계수를 계산하는 매서드

```python
import matplotlib.pyplot as plt
weather.hist(bins=30) # 데이터의 범위를 30개의 구간으로 나눔
plt.show()
```

```python
plt.scatter(
    weather['cloud'], 
    weather['sunshine'],
    color='green',
    alpha=0.5
)
plt.title('cloud-sunshine scatter')
plt.xlabel('cloud')
plt.ylabel('sunshine')
plt.show()
```

```python
plt.scatter(
    weather['cloud'], 
    weather['humi'],
    color='blue',
    alpha=0.5
)
plt.title('cloud-humi scatter')
plt.xlabel('cloud')
plt.ylabel('humi')
plt.show()
```
