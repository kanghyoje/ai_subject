```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('student.csv')
df.head()
```

```python
# 공부시간 분포 알아보기, 히스토그램
plt.hist(df['StudyHours'])
plt.xlabel('Study Hours')
plt.ylabel('Frequency')
plt.title('Study Hours Distribution')
plt.show()
```

```python
# 공부시간과 과제점수의 관계 : 산점도
plt.scatter(
    df['StudyHours'], 
    df['AssignmentScore']
)
plt.xlabel('Study Hours')
plt.ylabel('Assignment Score')
plt.title('Study Hours vs Assignment score')
plt.show()
```

```python
# 통과와 미통과 집단의 공부시간 평균 비교하기-막대그래프
result_mean = df.groupby('Result')['StudyHours'].mean()
```

```python
plt.bar(result_mean.index, result_mean.values)
plt.xlabel('Result')
plt.xticks([0,1])
plt.ylabel('Average Study Hours')
plt.title('Average Study Hours by Result')
plt.show()
```

```python
# 개별화, 문제 1
plt.scatter(
    df['Attendance'], 
    df['AssignmentScore']
)
plt.xlabel('Attendance')
plt.ylabel('Assignment Score')
plt.title('Attendance and Assignment score')
plt.show()
```

```python
# 개별화, 문제2
plt.scatter(
    df['StudyHours'], 
    df['SleepHours']
)
plt.xlabel('StudyHours')
plt.ylabel('SleepHours')
plt.title('StudyHours and SleepHours')
plt.show()
```
